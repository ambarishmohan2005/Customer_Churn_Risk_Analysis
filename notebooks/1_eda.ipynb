```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
# Load the dataset
df = pd.read_csv('data/Telco-Customer-Churn.csv')
print(f'Dataset shape: {df.shape}')
print(f'Columns: {df.columns.tolist()}')

# First few rows
df.head()
```

```python
# Check missing values
print('Missing values in columns:')
print(df.isnull().sum())

# TotalCharges contains blank spaces for new customers; convert to numeric
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'].replace(' ', np.nan), errors='coerce').fillna(0.0)
print(f'Dataset shape after TotalCharges numeric conversion: {df.shape}')
```

```python
# Encode target variable
df['Churn'] = df['Churn'].map({'Yes': 1, 'No': 0})

# Churn distribution
plt.figure(figsize=(8, 5))
sns.countplot(x='Churn', data=df)
plt.title('Churn Distribution')
plt.xticks([0, 1], ['No Churn', 'Churn'])
plt.show()
```

```python
# Encode categorical features for correlation inspection
df_corr = df.drop(columns=['customerID'], errors='ignore').copy()
cat_cols = df_corr.select_dtypes(include=['object']).columns
for col in cat_cols:
    df_corr[col] = df_corr[col].astype('category').cat.codes

# Correlation heatmap
plt.figure(figsize=(14, 10))
sns.heatmap(df_corr.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Feature Correlation Heatmap')
plt.tight_layout()
plt.show()
```

```python
# Key statistics
print('Churn rate:', round(df['Churn'].mean(), 4))
print('\nClass distribution:')
print(df['Churn'].value_counts())

# Observations
print('\nEDA Complete! Key observations:')
print('- High churn rate among month-to-month contracts')
print('- High churn rate among fiber optic internet users')
print('- Higher monthly charges strongly correlate with higher churn')
print('- High tenure customers show significantly lower churn')
```
